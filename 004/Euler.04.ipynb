```python
def isPalindromic(n):
    s = str(n)
    return s == s[::-1]
```

```python
def palindromicList():
    w = []
    for i in range (100, 999):
        for j in range (100, 999):
            n = i * j
            if(isPalindromic(n)):
                w.append(n)
    w.sort()
    return w
```
