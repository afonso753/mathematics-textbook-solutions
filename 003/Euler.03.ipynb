```python
def isPrime(n):
    i=2
    prime = True
    while(i<n and prime == True):
        if(n%i==0):
            prime = False
        else: i+=1
    return prime
            
```

```python
def largestPrime(n):
    i=2
    r=1
    while(i<=n):
        if(n%i==0 and isPrime(i)):
            r=i
            i+=1
        else:
            i+=1
    return r
    
```
